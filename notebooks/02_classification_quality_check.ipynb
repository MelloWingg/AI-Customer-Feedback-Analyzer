```python
import pandas as pd
```

```python
results_df = pd.read_csv("../data/processed/test_results_batch_1.csv")
```

```python
print("Dataset shape:", results_df.shape)

print("\nMissing values:")
print(results_df.isna().sum())

print("\nDuplicate review IDs:")
print(results_df["review_id"].duplicated().sum())

print("\nSentiment values:")
print(results_df["sentiment"].value_counts())

print("\nCategory values:")
print(results_df["category"].value_counts())

#print("\nProcessing status:")
#print(results_df["status"].value_counts())
```

```python
allowed_sentiments = [
    "Positive",
    "Neutral",
    "Negative"
]

allowed_categories = [
    "Delivery",
    "Product Quality",
    "Product Description",
    "Seller Service",
    "Packaging",
    "Refund & Returns",
    "Payment & Pricing",
    "Other"
]

print("Sentiment distribution:")
print(results_df["sentiment"].value_counts().to_string())

print("\nCategory distribution:")
print(results_df["category"].value_counts().to_string())

print("\nUnexpected sentiments:")
print(
    results_df[
        ~results_df["sentiment"].isin(allowed_sentiments)
    ]["sentiment"].unique()
)

print("\nUnexpected categories:")
print(
    results_df[
        ~results_df["category"].isin(allowed_categories)
    ]["category"].unique()
)
```

```python
print(results_df["review_id"].dtype)
print(results_df["review_id"].isna().sum())
```

```python
duplicate_ids = results_df["review_id"].value_counts()
duplicate_ids = duplicate_ids[duplicate_ids > 1].index

duplicates = results_df[results_df["review_id"].isin(duplicate_ids)]

display(duplicates.sort_values("review_id"))
```

```python
print("Number of duplicated IDs:", len(duplicate_ids))
print("Rows involved:", len(duplicates))
```

```python
print(duplicates.groupby("review_id").nunique())
```

```python
df = results_df.drop_duplicates(subset="review_id", keep="first").copy()

print("Dataset shape:", df.shape)
print("Duplicate IDs:", df["review_id"].duplicated().sum())
```

```python
df.to_csv(
    "../data/processed/classified_reviews_4000_clean.csv",
    index=False
)
```

```python
print(df.dtypes)

print("\nUnique values:")
print("Sentiment:", df["sentiment"].unique())
print("Category:", df["category"].unique())

print("\nReview score:")
print(df["review_score"].value_counts().sort_index())
```
